## **🦢 Generate the mjswan App**

```python
import mjswan

run_path = "ttktjmt-org/mjlab/dqxvf0eb"  # @param {type: "string"}
task_id = "Mjlab-Velocity-Flat-Anymal-C"

builder = mjswan.Builder.from_mjlab(
    task_id=task_id,
    run_path=run_path,
)

app = builder.build(output_dir="/content/mjswan/app")
```

## **🌐 Serve Visualization**

```python
app.launch(height=300)
```
